__Randomly deconnect electrodes or shake them__


```python
# get all positions as pairs, shuffle with fixed seed, take first XX , 
np.random.seed(13)
radius = np.linspace(0,6, 31)
degrees = np.linspace(180,272,24)
degrees.astype(int)
positions, meta_positions = create_trajectory_rdegbased(radius, degrees, factor=factor)
P = len(positions)
shuffle = np.arange(P)
np.random.shuffle(shuffle)
Pslice= P//10

sSavePath= sSavePathOriginal + "Phase5Disturbances/"
start_procedure(sciospec, setup)
sCurrPath = make_results_folder(bCreateResultsFolder=True, bSaveData=True, sSavePath=sSavePath)

# CHOOSE CORRECT PROBE
meta, i = MeaMeta(sCurrPath, "smallcylinder", [0,0], "none", "static", 0.7, 1.54), 0
#meta, i= MeaMeta(sCurrPath, "mediumcylinder", [0,0], "none", "static", 1.35, 5.73),1
#meta, i = MeaMeta(sCurrPath, "largecylinder", [0,0], "none", "static", 1.8, 10.18),2
#meta, i=  MeaMeta(sCurrPath, "smalltube", [0,0], "none", "low", 0.8, 1.99),3
#meta, i= MeaMeta(sCurrPath, "mediumtube", [0,0], "none", "low", 0.8, 3.96), 4
#meta, i= MeaMeta(sCurrPath, "largetube", [0,0], "none", "low", 0.8, 4.87),5

positions, meta_positions = positions[shuffle], meta_positions[shuffle]
posslice, meta_posslice = positions[i*Pslice:(i+1)*Pslice], meta_positions[i*Pslice:(i+1)*Pslice]

for iIdx, pos in tqdm(enumerate(posslice), total=len(posslice)):
    printer.move_to_absolute(pos)
    meta.set_position(posslice[iIdx])
    temp = printer.read_temperature()
    # for in jupyter print("\r", b, end="")
    print(f"Measuring with {meta.sProbetype} at {meta.piPositions}",end="\r")
    # With delay and automatic saving of meta data
    measurement_static_with_meta(sciospec, sCurrPath, meta, sleeptime=10,timeout=4, printer_temp=temp)
print(f"Measurement with Disturbances for {meta.sProbetype} finished.")

```
